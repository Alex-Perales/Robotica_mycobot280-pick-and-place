## Poses Clave del Ciclo

| Pose | Función |
|---|---|
| pose_inicial | Reposo seguro |
| pos_mira / pose_observar | Vista cenital del cubo |
| pos_agarrar | Descenso al objeto |
| pos_alto | Punto intermedio seguro (anti-colisión) |
| pos_mover | Giro por arriba al otro lado |
| pos_dejar | Punto de depósito |


```python
# ============================================================
# POSES CLAVE + POSE INICIAL
# ============================================================

pose_inicial   = [0.7,    -0.17,  -0.61,  -1.58,  -0.35, -44.29]
pose_observar  = [-53.87, -56.95, -7.38,  -31.28,  3.33, -27.86]
pose_agarrar   = [-52.55, -71.01, -7.38,  -31.11,  3.25, -27.77]
pose_alto      = [-52.55, -50.09, -6.85,  -31.11,  3.33, -27.94]
pose_mover     = [99.05,  -33.13, -48.51,  -7.47,  3.77, -44.82]
pose_depositar = [99.4,   -48.33, -48.6,   -6.94,  3.33, -44.38]

# Mover a pose inicial
mc.send_angles(pose_inicial, 30)
time.sleep(3)

print("Poses cargadas correctamente")
print("pose_inicial   :", pose_inicial)
print("pose_observar  :", pose_observar)
print("pose_agarrar   :", pose_agarrar)
print("pose_alto      :", pose_alto)
print("pose_mover     :", pose_mover)
print("pose_depositar :", pose_depositar)
```

## Un Ciclo Completo (Prueba Inicial)

Ejecuta una sola vez la secuencia: ir a mirar el cubo → bajar → cerrar gripper → subir → girar → depositar → soltar → volver.


```python
# ============================================================
# UN CICLO COMPLETO — Posiciones recalibradas
# ============================================================

from pymycobot.mycobot import MyCobot
import time

mc = MyCobot('/dev/ttyUSB0', 1000000)
mc.power_on()
time.sleep(1)

# Poses
pose_inicial = [0.7,    -0.17,  -0.61,  -1.58,   -0.35, -44.29]
pos_mira     = [-53.87, -56.95, -7.38,  -31.28,   3.33, -27.86]
pos_agarrar  = [-52.55, -71.01, -7.38,  -31.11,   3.25, -27.77]
pos_alto     = [-52.55, -50.09, -6.85,  -31.11,   3.33, -27.94]
pos_mover    = [99.05,  -33.13, -48.51,  -7.47,   3.77, -44.82]
pos_dejar    = [99.4,   -48.33, -48.6,   -6.94,   3.33, -44.38]

VELOCIDAD = 20

# Ir a pose inicial
print("Moviendo a pose inicial...")
mc.send_angles(pose_inicial, VELOCIDAD)
time.sleep(3)

# PASO 1: Ir a mirar el cubo
print("PASO 1 — Bajando a mirar el cubo...")
mc.send_angles(pos_mira, VELOCIDAD)
time.sleep(3)

# PASO 2: Bajar al cubo con gripper abierto
print("PASO 2 — Bajando al cubo...")
mc.set_gripper_value(100, 50)
time.sleep(1)
mc.send_angles(pos_agarrar, VELOCIDAD)
time.sleep(3)

# PASO 3: Cerrar gripper
print("PASO 3 — Cerrando gripper...")
mc.set_gripper_value(0, 50)
time.sleep(1)

# PASO 4: Subir alto
print("PASO 4 — Subiendo alto con el cubo...")
mc.send_angles(pos_alto, VELOCIDAD)
time.sleep(3)

# PASO 5: Girar al otro lado
print("PASO 5 — Girando por arriba al otro lado...")
mc.send_angles(pos_mover, VELOCIDAD)
time.sleep(5)

# PASO 6: Bajar al punto de deposito
print("PASO 6 — Bajando al punto de deposito...")
mc.send_angles(pos_dejar, VELOCIDAD)
time.sleep(3)

# PASO 7: Abrir gripper
print("PASO 7 — Abriendo gripper, soltando cubo...")
mc.set_gripper_value(100, 50)
time.sleep(2)

# Volver a pose inicial
print("Volviendo a pose inicial...")
mc.send_angles(pose_inicial, VELOCIDAD)
time.sleep(3)

print("LISTO — Ciclo completado")
```

## Cinco Ciclos Consecutivos

Ejecuta el ciclo 5 veces sin intervención humana, requisito de P5.
Velocidad reducida a 20% para garantizar repetibilidad.


```python
# ============================================================
# 5 CICLOS COMPLETOS — Posiciones recalibradas
# ============================================================
from pymycobot.mycobot import MyCobot
import time

mc = MyCobot('/dev/ttyUSB0', 1000000)
mc.power_on()
time.sleep(1)

# Poses
pose_inicial = [0.7,    -0.17,  -0.61,  -1.58,   -0.35, -44.29]
pos_mira     = [-53.87, -56.95, -7.38,  -31.28,   3.33, -27.86]
pos_agarrar  = [-52.55, -71.01, -7.38,  -31.11,   3.25, -27.77]
pos_alto     = [-52.55, -50.09, -6.85,  -31.11,   3.33, -27.94]
pos_mover    = [99.05,  -33.13, -48.51,  -7.47,   3.77, -44.82]
pos_dejar    = [99.4,   -48.33, -48.6,   -6.94,   3.33, -44.38]

VELOCIDAD = 20

# Ir a pose inicial
print("Moviendo a pose inicial...")
mc.send_angles(pose_inicial, VELOCIDAD)
time.sleep(3)

for ciclo in range(1, 6):
    print("\n" + "=" * 50)
    print(f"   CICLO {ciclo} de 5")
    print("=" * 50)

    # PASO 1: Ir a mirar el cubo
    print("PASO 1 — Bajando a mirar el cubo...")
    mc.send_angles(pos_mira, VELOCIDAD)
    time.sleep(3)

    # PASO 2: Bajar al cubo con gripper abierto
    print("PASO 2 — Bajando al cubo...")
    mc.set_gripper_value(100, 50)
    time.sleep(1)
    mc.send_angles(pos_agarrar, VELOCIDAD)
    time.sleep(3)

    # PASO 3: Cerrar gripper
    print("PASO 3 — Cerrando gripper...")
    mc.set_gripper_value(0, 50)
    time.sleep(1)

    # PASO 4: Subir alto
    print("PASO 4 — Subiendo alto con el cubo...")
    mc.send_angles(pos_alto, VELOCIDAD)
    time.sleep(3)

    # PASO 5: Girar al otro lado
    print("PASO 5 — Girando por arriba al otro lado...")
    mc.send_angles(pos_mover, VELOCIDAD)
    time.sleep(5)

    # PASO 6: Bajar al punto de deposito
    print("PASO 6 — Bajando al punto de deposito...")
    mc.send_angles(pos_dejar, VELOCIDAD)
    time.sleep(3)

    # PASO 7: Abrir gripper
    print("PASO 7 — Abriendo gripper, soltando cubo...")
    mc.set_gripper_value(100, 50)
    time.sleep(2)

    # Volver a pose inicial
    print("Volviendo a pose inicial...")
    mc.send_angles(pose_inicial, VELOCIDAD)
    time.sleep(3)

    print(f"CICLO {ciclo} COMPLETADO")

print("\n" + "=" * 50)
print("   5 CICLOS COMPLETADOS")
print("=" * 50)
```
